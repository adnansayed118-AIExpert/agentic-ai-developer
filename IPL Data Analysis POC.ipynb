```python
# how to add legend() in visualisation

plt.plot(Salary[0],c='g',ls='--',marker='*',ms=7,label=[0]) #markers size 'ms'
plt.plot(Salary[1],c='k',ls=':',marker='^',ms=8,label=[1])
plt.plot(Salary[2],c='b',ls=':',marker='p',ms=9,label=[2])
plt.plot(Salary[3],c='m',ls=':',marker='o',ms=6,label=[3])
plt.legend(loc='lower right', bbox_to_anchor=(0.5,1))
plt.xticks(list(range(0,10)), Seasons,rotation='vertical')
plt.show
```
